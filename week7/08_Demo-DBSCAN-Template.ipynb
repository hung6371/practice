```python
import numpy as np
import matplotlib
from matplotlib import pyplot as plt

np.set_printoptions(precision=3, suppress=True)
```

## 1. Dữ liệu giả định
(Cùng dữ liệu giả với `03_Demo-kMeans.ipynb`)

```python
np.random.seed(1)
x1,y1 = np.random.normal(loc=2,scale=1,size=50), np.random.normal(loc=5,scale=1,size=50)
x2,y2 = np.random.normal(loc=5,scale=1,size=75), np.random.normal(loc=1,scale=1,size=75)
x3,y3 = np.random.normal(loc=8,scale=1,size=100), np.random.normal(loc=7,scale=1,size=100)
```

```python
fig = plt.figure(figsize=(6,6))
plt.xlim([-2,12])
plt.ylim([-2,12])

plt.scatter(x1, y1)
plt.scatter(x2, y2)
plt.scatter(x3, y3)
plt.show()
```

### Ghép nối dữ liệu lại thành 1 tập

```python
dat1 = np.stack([x1,y1]).T
dat1.shape
```

```python
dat2 = np.stack([x2,y2]).T
dat3 = np.stack([x3,y3]).T
```

```python
dat = np.vstack([dat1, dat2, dat3])
dat.shape
```

### Sắp ngẫu nhiên các dòng

```python
np.random.seed(2)
np.random.shuffle(dat)
```

### Kiểm tra lại với biểu đồ scatter

```python
fig = plt.figure(figsize=(6,6))
plt.xlim([-2,12])
plt.ylim([-2,12])

plt.scatter(dat[:,0], dat[:,1], color='y')
plt.show()
```

## 2. Thực hiện phân cụm bằng DBSCAN

**=== Import lớp DBSCAN từ thư viện sklearn ===**

```python
from sklearn.cluster import DBSCAN
```

### 2.1. Khởi tạo model và fit vào dữ liệu

```python
# Giả sử chọn epsilon = 1, minPoints = 5
dbscan = DBSCAN(eps=1, min_samples=5)
```

**=== Thực hiện fit vào dữ liệu ===**

```python
dbscan.fit(dat)
```

### 2.2. Kiểm tra kết quả các cluster tìm được

**=== Lấy ra nhãn các clusters ===**

```python
cluster = dbscan.labels_
```

```python
cluster
```

**=== Đếm xem mỗi cluster có bao nhiêu điểm ===**

```python
nhan, dem = np.unique(cluster, return_counts=True)
for n, d in zip(nhan, dem):
    print('nhãn', n, ':', d, 'điểm')
```

**=== Hiển thị lên biểu đồ ===**

```python
fig = plt.figure(figsize=(6,6))
plt.xlim([-2,12])
plt.ylim([-2,12])

for n in np.unique(cluster):
    pts = dat[cluster == n]
    if n == -1:
        plt.scatter(pts[:,0], pts[:,1], color='k', marker='x', label='noise')
    else:
        plt.scatter(pts[:,0], pts[:,1], label=f'cluster {n}')
plt.legend()

plt.show()
```

### 2.3. Tính hiệu suất phân cụm

**=== Thử tính với hàm built-in của Sklearn ===**

```python
from sklearn.metrics import silhouette_score
```

```python
print('Silhouette (giữ noise):', silhouette_score(dat, cluster))
```

**=== Tính lại với các cụm không chứa điểm noise ===**

```python
import pandas as pd
```

```python
# Bỏ đi các samples bị xác định là noise
mask = cluster != -1
dat2 = dat[mask]
cluster2 = cluster[mask]
print(dat2.shape, np.unique(cluster2))
print('Silhouette sklearn (bỏ noise):', silhouette_score(dat2, cluster2))
```

```python
# Tạo một list rỗng để chứa hệ số Silhouette của mỗi sample
s_list = []

for i, sample in enumerate(dat2):
    label = cluster2[i]

    # Lấy ra các samples khác và nhãn khác
    other_samples = np.delete(dat2, i, axis=0)
    other_labels = np.delete(cluster2, i)

    # Tính khoảng cách từ sample đến tất cả các other_samples
    distances = np.sqrt(np.sum((other_samples - sample)**2, axis=1))

    # Tạo dataframe với 2 cột 'distance' và 'cluster' để dùng groupby tính mean()
    d = pd.DataFrame({'distance': distances, 'cluster': other_labels})
    mean_d = d.groupby('cluster')['distance'].mean()

    # Lấy ra giá trị p, q tương ứng
    p = mean_d[label]                  # khoảng cách trung bình tới cụm của chính nó
    q = mean_d.drop(label).min()       # khoảng cách trung bình tới cụm gần nhất còn lại

    # Tính ra hệ số Silhouette cho sample
    s_list.append((q - p) / max(p, q))
```

```python
# Tính trung bình hệ số Silhouette của tất cả các samples
s_mean = np.mean(s_list)
print('Silhouette tự tính (bỏ noise):', s_mean)
print('Khớp với sklearn:', np.isclose(s_mean, silhouette_score(dat2, cluster2)))
```

**==> Thử thực hiện lại mà không bỏ giá trị noise?**

```python
# Không bỏ noise: dùng cả dat và cluster, nhãn -1 được xem như một "cụm" riêng
s_list_all = []

for i, sample in enumerate(dat):
    label = cluster[i]
    other_samples = np.delete(dat, i, axis=0)
    other_labels = np.delete(cluster, i)
    distances = np.sqrt(np.sum((other_samples - sample)**2, axis=1))
    d = pd.DataFrame({'distance': distances, 'cluster': other_labels})
    mean_d = d.groupby('cluster')['distance'].mean()
    p = mean_d[label]
    q = mean_d.drop(label).min()
    s_list_all.append((q - p) / max(p, q))

print('Silhouette tự tính (giữ noise):', np.mean(s_list_all))
print('Khớp với sklearn:', np.isclose(np.mean(s_list_all), silhouette_score(dat, cluster)))

# Hệ số của riêng các điểm noise
print('Silhouette trung bình của các điểm noise:', np.mean(np.array(s_list_all)[cluster == -1]))
```

**Nhận xét:** với `eps=1`, `minPoints=5` DBSCAN tìm được 3 cụm (98, 73, 47 điểm) và 7 điểm noise (khoảng 3.1% trong 225 điểm).

- Silhouette khi **giữ noise** (nhãn `-1` xem như một cụm): 0.639.
- Silhouette khi **bỏ noise**: 0.685, cao hơn.
- Các điểm noise có hệ số Silhouette trung bình âm (khoảng -0.574): chúng nằm ở vùng thưa giữa các cụm, gần cụm khác hơn là "cụm -1". Chúng kéo điểm trung bình xuống.

Noise không thuộc cụm nào, nên khi đánh giá chất lượng các cụm người ta thường bỏ noise, nhưng vẫn nên báo cáo kèm tỉ lệ điểm bị loại.
